```python
hbb = "ATGGTGCATCTGACTCCTGAGGAGAAGTCTGCCGTTACTGCCCTGTGGGGCAAGGTGAAC"
def gc_content(my_sequence):
    """ It will calculate the GC content of your sequence """
    num_gc = my_sequence.count("G")+my_sequence.count("C")
    return ((num_gc)/(len(my_sequence)))*100
print(gc_content(hbb))
```

```python
with open("gc_content_report.txt","w") as fh:
    fh.write("GC content of HBB seq reported ")
    fh.write(str(round(gc_content(hbb),2)))
with open("gc_content_report.txt") as fh:
    print(fh.read())
```
