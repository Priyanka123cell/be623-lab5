10. Read a3.fasta line by line in Python. Collect every header line into a list, then print how many headers you found,             followed by the header lines themselves, one per line.

```python
headers = []   #initializing empty list called headers
for line in open("a3.fasta"):   # for loop used and open(a3.fasta) opens and accesses the file a3.fasta
    if line.startswith(">"):   # if condition used to check if line starts with header symbol
        headers.append(line.strip())   # if the line starts with > then that line has \n (newline character) at the end and \n is stripped using line.strip() and appended to the list headers so that each header line is added to the list
print(len(headers))       #prints lenght of the headers list to count number of headers
for header in headers:    # for loop used to print each header from th elist headers line by line
    print(header)         # prints the header lines line by line
```
